```python
import pandas as pd
import os
from urllib.parse import urlparse, parse_qs
```

## Plaftorm Tokopedia

```python
# list file
files = ['laptop_baru_tokopedia.csv','laptop_bekas_tokopedia.csv']

# fungsi extract brand
def extract_brand(url):
    try:
        query = urlparse(url).query
        params = parse_qs(query)

        if 'q' in params:
            keyword = params['q'][0]
            words = keyword.lower().split()
            brand = words[-1]
            return brand.title()
    except:
        return None

# fungsi ambil info dari nama file
def extract_info_from_filename(filepath):
    filename = os.path.basename(filepath)
    name = filename.replace('.csv', '')
    parts = name.split('_')

    kategori = parts[0].title()
    kondisi = parts[1].title()
    platform = parts[2].title()

    return kategori, kondisi, platform


df_list = []

for file in files:
    df = pd.read_csv(file)

    # ambil info file
    kategori, kondisi, platform = extract_info_from_filename(file)

    # tambah kolom baru
    df['brand'] = df['web_scraper_start_url'].apply(extract_brand)
    df['kategori'] = kategori
    df['kondisi'] = kondisi
    df['platform'] = platform

    # rename kolom sesuai permintaan
    df = df.rename(columns={
        'nama produk': 'nama_produk',
        'jumlah terjual': 'jumlah_terjual',
        'lokasi penjual': 'lokasi_penjual'
    })

    # drop kolom tidak perlu termasuk nama toko
    df = df.drop(columns=['web_scraper_order', 'web_scraper_start_url', 'detail produk', 'nama toko'], errors='ignore')

    df_list.append(df)

# gabungkan semua
df_tokopedia = pd.concat(df_list, ignore_index=True)
```

## Platform Blibli


```python
import pandas as pd
import glob
import os

# ======================
# SOURCE FILE
# ======================
folder_path = "/content"
files = glob.glob(f"{folder_path}/*.csv")

# filter hanya file blibli (opsional kalau namanya spesifik)
files = [
    f for f in files
    if 'clean' not in f.lower() and 'blibli' in f.lower()
]

# ======================
# EXTRACT METADATA
# ======================
def extract_info_from_filename(filepath):
    filename = os.path.basename(filepath)
    name = filename.replace('.csv', '')
    parts = name.split('_')

    kategori = parts[0].title() if len(parts) > 0 else ""
    brand = parts[1].title() if len(parts) > 1 else ""
    platform = parts[2].title() if len(parts) > 2 else ""

    return kategori, brand, platform


# ======================
# PROCESS FILES
# ======================
df_list = []

print(f"📁 Total file ditemukan: {len(files)}\n")

for file in files:

    try:
        print(f"📄 Memproses → {os.path.basename(file)}")

        df = pd.read_csv(file)

        kategori, brand, platform = extract_info_from_filename(file)

        # tambah metadata
        df['kategori'] = kategori
        df['brand'] = brand
        df['platform'] = platform

        # rename kolom
        df = df.rename(columns={
            'nama produk': 'nama_produk',
            'jumlah terjual': 'jumlah_terjual',
            'lokasi penjual': 'lokasi_penjual'
        })

        # drop kolom tidak perlu
        df = df.drop(columns=[
            'web_scraper_order',
            'web_scraper_start_url',
            'nama toko'
        ], errors='ignore')

        df_list.append(df)

    except Exception as e:
        print(f"❌ Gagal proses file: {file}")
        print(e)


# ======================
# FINAL DATAFRAME
# ======================
df_blibli = pd.concat(df_list, ignore_index=True)

print("\n🎉 SELESAI!")
print(f"✅ Total data: {len(df_blibli):,}")
```

## Shopee

```python
import pandas as pd
import glob
import re
from pathlib import Path

# ======================
# SOURCE FILES
# ======================
INPUT_PATTERN = "/content/*.csv"
OUTPUT_FILE = "data_gabungan_shopee.csv"

# ======================
# PARSE CONTAINER
# ======================
def parse_shopee_container(text):

    if pd.isna(text):
        return None

    text = str(text).strip()

    if text == "":
        return None

    lines = [line.strip() for line in text.split('\n') if line.strip()]

    if len(lines) == 0:
        return None

    nama_produk = lines[0]

    harga = ""
    rating = ""
    jumlah_terjual = ""
    lokasi_penjual = ""

    for i, line in enumerate(lines[1:], start=1):

        clean_line = line.strip()

        # HARGA
        if not harga:

            match = re.search(r'Rp\s*[\d\.\,]+', clean_line, re.I)
            if match:
                harga = match.group().replace(" ", "")
                continue

            if clean_line.lower() == "rp" and i + 1 < len(lines):
                next_line = lines[i + 1].strip()
                if re.search(r'\d', next_line):
                    harga = "Rp" + next_line

        # RATING
        if not rating:
            rating_match = re.match(r'^([1-5](\.\d+)?)$', clean_line)
            if rating_match:
                rating = rating_match.group(1)

        # JUMLAH TERJUAL
        if not jumlah_terjual:
            if re.search(r'terjual', clean_line, re.I):
                jumlah_terjual = clean_line

        # LOKASI
        if not lokasi_penjual:
            lokasi_keywords = [
                'jakarta','bandung','surabaya','bekasi',
                'tangerang','depok','medan','semarang',
                'makassar','yogyakarta','malang','bogor',
                'banten','bali','indonesia'
            ]

            if any(k in clean_line.lower() for k in lokasi_keywords):
                lokasi_penjual = clean_line

    return {
        "nama_produk": nama_produk,
        "harga": harga,
        "rating": rating,
        "jumlah_terjual": jumlah_terjual,
        "lokasi_penjual": lokasi_penjual
    }


# ======================
# DETECT METADATA
# ======================
def detect_metadata(file_name):

    fname = Path(file_name).stem.lower()
    fname = re.sub(r'\(\d+\)', '', fname)

    parts = fname.split('_')

    kategori = parts[0].capitalize() if len(parts) > 0 else ""
    brand = parts[1].capitalize() if len(parts) > 1 else ""
    kondisi = parts[2].capitalize() if len(parts) > 2 else ""
    platform = parts[3].capitalize() if len(parts) > 3 else ""

    return kategori, brand, kondisi, platform


# ======================
# MAIN PROCESS
# ======================
print("🔍 Mencari file CSV di /content ...\n")

# FILTER SHOPEE ONLY
files_found = [
    f for f in glob.glob(INPUT_PATTERN)
    if 'shopee' in f.lower() and 'clean' not in f.lower()
]

all_records = []

print(f"📁 Total file Shopee ditemukan: {len(files_found)}\n")

for file_path in sorted(files_found):

    file_name = Path(file_path).name
    print(f"📄 Memproses → {file_name}")

    kategori, brand, kondisi, platform = detect_metadata(file_name)

    try:
        df = pd.read_csv(file_path, low_memory=False)
    except Exception as e:
        print(f"❌ Gagal membaca: {file_name}")
        print(e)
        continue

    container_col = None
    possible_cols = ['_container', 'container', 'data', 'content']

    for col in possible_cols:
        if col in df.columns:
            container_col = col
            break

    if not container_col:
        print("⚠️ Kolom container tidak ditemukan")
        continue

    for _, row in df.iterrows():

        parsed = parse_shopee_container(row.get(container_col, ""))

        if parsed:
            parsed['kategori'] = kategori
            parsed['brand'] = brand
            parsed['kondisi'] = kondisi
            parsed['platform'] = platform

            all_records.append(parsed)


# ======================
# FINAL DATAFRAME
# ======================
result = pd.DataFrame(all_records)

if not result.empty:
    result = result[
        [
            'nama_produk',
            'harga',
            'rating',
            'jumlah_terjual',
            'lokasi_penjual',
            'kategori',
            'brand',
            'kondisi',
            'platform'
        ]
    ]

    result = result.drop_duplicates()

# ======================
# OUTPUT
# ======================
print("\n🎉 SELESAI!")
print(f"✅ Total data Shopee: {len(result):,}")

result.to_csv(OUTPUT_FILE, index=False, encoding='utf-8-sig')
print(f"💾 Saved: {OUTPUT_FILE}")
```
